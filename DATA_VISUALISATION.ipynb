```python
from IPython.display import display
import ipywidgets as widgets
import pandas as pd
from io import BytesIO

upload = widgets.FileUpload(
    accept='.csv',
    multiple=False
)

display(upload)
```

```python
# Get the uploaded file
filename = list(upload.value.keys())[0]

# Read the CSV
content = upload.value[filename]['content']

df = pd.read_csv(BytesIO(content))

# Display the dataset
print("Dataset loaded successfully!")
print(df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. BAR CHART
plt.figure(figsize=(10, 5))
plt.bar(df["Student_Name"], df["Marks"])
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.title("Marks of Students")
plt.xticks(rotation=45)
plt.show()


# 2. LINE CHART
plt.figure(figsize=(10, 5))
plt.plot(df["Student_Name"], df["Marks"], marker='o')
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()


# 3. SCATTER PLOT
plt.figure(figsize=(8, 5))
plt.scatter(df["Study_Hours"], df["Marks"])
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.grid(True)
plt.show()


# 4. PIE CHART
categories = pd.cut(
    df["Marks"],
    bins=[0, 50, 70, 85, 100],
    labels=["Below 50", "50-70", "71-85", "86-100"]
)

category_count = categories.value_counts()

plt.figure(figsize=(7, 7))
plt.pie(
    category_count,
    labels=category_count.index,
    autopct="%1.1f%%"
)
plt.title("Students by Marks Category")
plt.show()


# 5. HISTOGRAM
plt.figure(figsize=(8, 5))
plt.hist(df["Marks"], bins=6)
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Marks")
plt.show()


# 6. BOX PLOT
plt.figure(figsize=(7, 5))
plt.boxplot(df["Marks"])
plt.ylabel("Marks")
plt.title("Box Plot of Marks")
plt.show()


# 7. HEATMAP
numeric_columns = [
    "Study_Hours",
    "Attendance_Percent",
    "Screen_Time_Hours",
    "Sleep_Hours",
    "Marks"
]

correlation = df[numeric_columns].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
